```python
## Basic CPW Design ##
import numpy as np

'''
    Limitations (CPW)
        h > (2-3)w & s
        h/s >= 1
'''

width = 75e-6
gap = 75e-6
height = 200e-6
rel = 9.4 ##Al2O3
thickness = 10e-6
Res = 2.44e-8   ##gold
max_f = 100e9

def k(w, s):
    # w = width, s = spacing from gnd
    return (w)/(w + 2*s)

def kp(w, s):
    return np.sqrt(1-k(w, s)**2)

def KFactor(w, s):
    const = k(w, s)
    if (const >= 0 and const <= 0.707):
        result = (1/np.pi)*np.log(2 * (1 + np.sqrt(k(w, s))) / (1 - np.sqrt(k(w, s))))
        result = 1/result
    elif (const > 0.707 and const <= 1):
        result = (1/np.pi)*np.log(2 * (1 + np.sqrt(kp(w, s))) / (1 - np.sqrt(kp(w, s))))
    return result

def Eps(er, h, s, w):
    const = k(w, s)
    result = 0.5*(er + 1)*(np.tanh(1.785 * np.log(h/s) + 1.75) + (const*s/h)*(0.04 - 0.7*const + 0.01*(1 - 0.1*er)*(0.25 + const)))
    return result

def X(eps, w, s):
    return 30 * np.pi * KFactor(w, s) / np.sqrt(eps)

def R(w, t, OM):
    return OM/(w*t)

def Z(r, x):
    return np.sqrt(r**2 + x**2)

def Pitch(f, eps):
    ## c = lambda*freq, lambda = c/freq 
    c = 3e8
    return round((1/20)*(c/(f*np.sqrt(eps))), 8)

print("Co-planer Transmission Line Calculator")
print("")

check = height/gap
text = "height/gap is: " + str(check) + ". Needs to be above 2-3"
print(text)

check = height/width
text = "height/width is: " + str(check) + ". Needs to be above 2-3"
print(text)

check = k(width, gap)
text = "k value is: " + str(check) + ". Needs to be between 0-1"
print(text)

perm = Eps(rel, height, gap, width)
text = "Permitivity is: " + str(perm)
print(text)

Reactance = X(perm, width, gap)
text = "Reactive impedance is: " + str(Reactance)
print(text)

Resistance = R(width, thickness, Res)
text = "Resistance is: " + str(Resistance)
print(text)

Impedance = Z(Resistance, Reactance)
text = "Impedance is: " + str(Impedance)
print(text)

text = "for a frequency of: " + str(max_f/1e9) + 'GHz, use a via pitch of: ' + str(Pitch(max_f, perm)*1e6) + " to " + str(2*Pitch(max_f, perm)*1e6) + 'um'
print(text)
```

```python
## Basic Microstrip Design ##

import numpy as np

'''
    Limitations:
    0.01 < w/h < 100
'''

height = 20e-9
width = 20e-9
rel = 9.34
thickness = 10e-9

Res = 2.44e-8 ##gold##
max_f = 100e9

def F1(h, w):
    return 6 + (2*np.pi - 6)*np.exp(-(30.66*h/w)**0.7528)

def Zo(F, h, w):
    return 60*np.log(F*h/w + np.sqrt(1 + (2*h/w)**2))

def Z(z, ee):
    return z/np.sqrt(ee)

def eps_e(a, b, h, w, er):
    return (er + 1)/2 + (er - 1) / 2 * (1 + 10*h/w)**(-a*b)

def B(er):
    return 0.564*((er - 0.9)/(er + 3))**0.053

def A(w, h):
    u = w/h
    result = 1 + (1/49)*np.log((u**4 + (u/52)**2)/(u**4 + 0.432))
    result = result + (1/18.7)*np.log(1 + (u/18.1)**3)
    return result

def R(w, t, OM):
    return OM/(w*t)

def Im(R, X):
    return np.sqrt(R**2 + X**2)

def Pitch(f, eps):
    ## c = lambda*freq, lambda = c/freq 
    c = 3e8
    return round((1/20)*(c/(f*np.sqrt(eps))), 8)

print("Microstrip calculator")
print("")

check1 = width/height
text = "The following is the width/height: " + str(check1) + ", is needs be between 0.01 and 100"
print(text)

perm = eps_e(A(width, height), B(rel), height, width, rel)
text = "The effective permitivity is: " + str(perm)
print(text)

Reactance = Z(Zo(F1(height, width), height, width), perm)
text = "The reactance of the trace is: " + str(Reactance)
print(text)

Resistance = R(width, thickness, Res)
text = "The resistance of the trace is: " + str(Resistance)
print(text)

Impedance = Im(Resistance, Reactance)
text = "The impedance of the trace is: " + str(Impedance)
print(text)
```
