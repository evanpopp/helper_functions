```python
## Grating Coupler

import numpy as np

WL = 750e-9
m = 1

ed = 1
em = .174 + 4.86j    #index of gold at 1000nm


theta = 14.5 *np.pi/180
k = 2*np.pi/WL

km = 2*np.pi/WL*np.sqrt(em*ed/(em + ed))

P = 2*np.pi*m/(np.real(km) - k*ed*np.sin(theta))

print(P/WL)
#1.35?
```

```python
## E-field per shot ##

import numpy as np

e = 8.854e-12
c = 3e8

Pave = 10
t = 120e-15
num = 80e6
time = t*num
P = Pave/time
## other calculator ##
##P = 3914321


Dia = 1.1e-3
A = np.pi*(Dia/2)**2
I = P/A

#other calculator#
#I = 4118899236182 W/mm^2

E = np.sqrt(2*I/(e*c))

##print(P)
print(I)
print(E)
print(E*1e-9)

gap = 21.4e-9
V = E*gap
print(V)
# Need ~0.3V, Might be loss?

Eneed = 0.3/gap
print(Eneed*10**-9)
## Basically need 0.014V/nm at gap
```
